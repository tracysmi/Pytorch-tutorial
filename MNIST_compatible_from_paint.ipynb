```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.colors as mcolors
%matplotlib inline

import os
print("Current working directory:", os.getcwd())

# List all files and directories in the current directory
files_and_dirs = os.listdir()

print("Files and directories in the current directory:")
for file_or_dir in files_and_dirs:
    print(file_or_dir)
```

```python
from PIL import Image

# Load the image you saved from Paint
img = Image.open('five_280.png')

# Convert the image to grayscale ('L' mode)
img = img.convert('L')
```

```python
print(img.size)  # Should print (28, 28)
print(img.mode)  # Should print 'L' for grayscale
```

```python
# Resize the image to 28x28 pixels using the LANCZOS resampling algorithm
img_resized = img.resize((28, 28), Image.Resampling.LANCZOS)
print(img_resized.size)  # Should print (28, 28)
print(img_resized.mode)  # Should print 'L' for grayscale
```

```python
# Create a colormap that maps 0 to white and higher values to black
cmap = mcolors.LinearSegmentedColormap.from_list("mycmap", ["black", "white"])
plt.imshow(img_resized,cmap=cmap)
```

```python
# Save the converted image
img_resized.save('five_28.png')
```
